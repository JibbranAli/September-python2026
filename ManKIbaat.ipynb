```python
expense=[

    { "date": "24-09-2026", "cat" : "food", "expense": 250 },
    { "date": "24-09-2026", "cat" : "travel", "expense": 200 },
    { "date": "24-09-2026", "cat" : "food", "expense": 550 },
    { "date": "24-09-2026", "cat" : "helth", "expense": 1000 }

    
]
```

```python
key = input("Enter Category : ").lower()

total = 0 

for  entry in expense:
    if entry["cat"].lower() == key:
        total += entry["expense"]

            
print(f"Total expense on {key} is {total}")
```
