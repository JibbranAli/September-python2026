# STOP WATCH 

```python
print("STOP WATCH STARTED ")

timer = int(input("Enter stopwatch TImer : "))
timer += 1
for seconds in range(1,timer):
    print("ELAPSED ",  seconds , " Seconds")
    time.sleep(1)
print("STOP WATCH STOPPED")
```

```python
print("STOP WATCH STARTED ")

start = int(input("Start From seconds: "))
end = int(input ("Stop at seconds : "))
for total_seconds in range(start , end + 1 ):
    minutes = total_seconds // 60 
    seconds = total_seconds % 60 
    
    print(f"{minutes:02d}:{seconds:02d}")
    time.sleep(1)
print("stopwatch Stopped")
```
