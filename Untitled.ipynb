```python
def total_expense():
    total=0
    with open ("expense.txt", "r") as file:
        line=file.readlines()
        for line in line:
            data=line.strip().split(",")
            amount=float(data[1])
            total = total + amount
    print("Total Expense is : ", total)
```

```python
total_expense()
```

```python
def hightest_amount():
    highest=0
    highest_category= ""

    with open ("expense.txt", "r") as file:
        line=file.readlines()
        for line in line:
            category,amount=line.strip().split(",")
            amount = float(amount)

            if amount > highest:
                highest = amount
                highest_category = category
    print("Highest Expense is " , highest_category )
    print("Amount", highest)
```

```python
hightest_amount()
```

```python
def count_number():
    count = 0
    with open ("expense.txt", "r") as file:
        line=file.readlines()
        for line in line:
            count += 1
    print("Toatal Transansactions ", count)
```
