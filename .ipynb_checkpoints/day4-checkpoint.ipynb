```python
employee = {}
```

```python
type(employee)
```

```python
employee["name"] = input("Enter your name : ")
employee["department"] = input("Enter your deparment : ")
employee["salary"] = float(input("Enter your salary In LPA : "))


print(f"Name : {employee['name']} ")
print(f"Deparment : {employee['department']}")

print(f"Salary in LPA : {employee['salary']}")



if employee['salary'] >=  5 :
    print("Salary Category : High")
elif employee['salary'] >= 3:
    print("Salary Category : Moderate")
else:
    print("Salary Category : LOW")
    
```
