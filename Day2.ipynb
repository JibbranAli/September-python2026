```python
#Multiple Conditions

name = input ("Enter your name :")
marks = int(input ("Enter your Marks : "))
if marks >=80:
    print(f"Congratulations !!! {name} , You GOT A GRADE")
elif marks >=60:
    print(f"Keep it up {name}, YOU GOT B GRADE")
elif marks >=40 :
    print(f"{name} You got C Grade , You need to Focus more on study .")
else:
    print(f"{name} You are fail You need to improve")
```
