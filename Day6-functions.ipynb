```python
def check_length (password):
    return len(password) >= 8 
```

```python
check_length("abc111111")
```

```python
def has_uppercase(password):
    for char in password:
        if char.isupper():
            return True
    return False
```

```python
has_uppercase("ajdjaj")
```

```python
def has_lowercase(password):
    for char in password:
        if char.islower():
            return True
    return False
```

```python
has_lowercase("AAAdsadasdas")
```

```python
def has_digit(password):
    for char in password:
        if char.isdigit():
            return True
    return False
```

```python
has_digit("adasds1111")
```

```python
@#!&*^%?/"{}()
```

```python
def has_special(password):
    special = "!@#$%^&*"

    for char in password:
        if char in special:
            return True
    return False
```

```python
has_special("abcha@")
```

```python
def security_score(password):
    score = 0 
    if check_length(password):
        score += 1
    if has_uppercase(password):
        score += 1
    if has_lowercase(password):
        score += 1
    if has_digit(password):
        score += 1
    if has_special(password):
        score +=1 
    return score
```

```python

security_score("Redhat@1")
```

```python
def security_level(score):
    if score == 5:
        return "VERY STRONG"
    elif score >=4:
        return "Strong"
    elif score >=3:
        return "Medium"
    elif score >=2:
        return "Weak"
    else:
        return "Very Weak "
```

```python
def password_report(password):
    score = security_score(password)

    print("""
    
            Passowrd Security LAB 
    """)

    print(f"Legth : ", len(password))

    print("LOWERCASE : " , 
          "YES" if has_lowercase (password) else "NO")

    print("UPPERCASE :" , 
          "YES" if has_uppercase (password) else "NO")

    print ("DIGIT : ", 
           "YES" if has_digit (password) else "NO")

    print("SPECIAL : " , 
          "YES" if has_special (password) else "NO")

    print("--------------------------------------------")

    print("SECURITY SCORE : ", score ,"/5")
    print("SECURITY LEVEL : ", security_level(score))
```

```python
while True:
    print("1 .Generate Password Report")
    print("2. CLOSE ")
    choice = input("ENter Your choice : ")
    if choice == "1":
        password = input("ENTER PASSWORD : ")
        password_report(password)

    elif choice == "2":
        break 
    else: 
        print("Enter valid CHoice")
```
